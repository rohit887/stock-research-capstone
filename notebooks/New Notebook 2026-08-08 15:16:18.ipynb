```python
from databricks.sdk import WorkspaceClient
w = WorkspaceClient()

# create the scope (ignore error if it already exists)
try:
    w.secrets.create_scope(scope="capstone")
except Exception as e:
    print("scope:", e)

w.secrets.put_secret(scope="capstone", key="massive_api_key",
                     string_value="3JsyOdjAxZuG4bsqSYnkXtU9il1bIdPa")

# verify (this lists keys only, never values)
print([s.key for s in w.secrets.list_secrets(scope="capstone")])
# -> ['massive_api_key']
```
